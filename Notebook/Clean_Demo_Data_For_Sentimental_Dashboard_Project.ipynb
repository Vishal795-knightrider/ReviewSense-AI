```python
import pandas as pd
```

```python
df = pd.read_csv('/content/imdb_dataset.csv', engine='python', on_bad_lines='skip')
```

```python
df_clean = df[['review', 'sentiment']].copy()
df_clean.rename(columns={'review':'comment'}, inplace=True)
```

```python
import re

def clean_text(text):
    text = re.sub(r'<.*?>', '', text)
    text = text.replace('\n', ' ').strip()
    return text

df_clean['comment'] = df_clean['comment'].apply(clean_text)
```

```python
df_clean['sentiment'] = df_clean['sentiment'].str.upper()
```

```python
df_clean.to_csv("comments.csv", index=False)
```
